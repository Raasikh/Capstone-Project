```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
import matplotlib.pyplot as plt


# Filter the last 3 years for training (2018-2020) and leave 2021/22 for prediction
train_data = df2[df2['Years'] < 2021]['Offered place on the waiting list']
test_data = df2[df2['Years'] >= 2021]['Offered place on the waiting list']

# Scale the data
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train_data = scaler.fit_transform(train_data.values.reshape(-1, 1))
scaled_test_data = scaler.fit_transform(test_data.values.reshape(-1, 1))

def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(seq_length, len(data)):
        X.append(data[i-seq_length:i, 0])  # Use previous seq_length years as input
        y.append(data[i, 0])  # Next year as the target
    return np.array(X), np.array(y)

# Set sequence length
seq_length = 3
X_train, y_train = create_sequences(scaled_train_data, seq_length)

# Reshape the data to fit LSTM model (samples, time steps, features)
X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))

# LSTM Model
model = Sequential()
model.add(LSTM(units=50, return_sequences=False, input_shape=(X_train.shape[1], 1)))
model.add(Dense(units=1))

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')

# Train the model
model.fit(X_train, y_train, epochs=100, batch_size=64)

# Prepare test data (use last 4 years to predict 2022)
last_3_years = scaled_train_data[-seq_length:]  # Get the last 3 years from training data
last_3_years = last_3_years.reshape((1, seq_length, 1))  # Reshape for LSTM input

# Predict 2022/21 value
predicted_scaled_2022 = model.predict(last_3_years)

# Invert scaling to get back to the original value
predicted_2022 = scaler.inverse_transform(predicted_scaled_2022)

# Print the actual and predicted values for 2022
actual_2022 = test_data.values[0]  # Actual 2022 value from the data
print(f"Actual value for 2022: {actual_2022}")
print(f"Predicted value for 2022: {predicted_2022[0][0]}")


```
